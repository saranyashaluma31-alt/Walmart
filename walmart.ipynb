```python
import pandas as pd

df = pd.read_excel(
    r"C:\Users\admin\Downloads\Amazon 2_Raw.xlsx",
    engine="openpyxl"
)

print(df.head())
```

```python
#Shape → Rows and columns
#Columns → Available fields
#Info → Data types
#Describe → Statistics
```

```python
print(df.shape)
print(df.columns)
print(df.info)
print(df.describe)
```

```python
#Find empty values in the dataset.
print(df.isnull().sum())
```

```python
#Remove Missing Values
print(df.fillna(0, inplace=True))
```

```python
#Check Duplicate Records
print(df.drop_duplicates(inplace=True))
```

```python
print(list(df.columns))
```

```python
print(df.columns.tolist())
```

```python
df.columns = df.columns.str.strip()
print(df.columns.tolist())
```

```python
#Total Sales, Quantity, Profit
print("Total Sales:", df['Sales'].sum())

print("Total Quantity:", df['Quantity'].sum())

print("Total Profit:", df['Profit'].sum())
```

```python
#Top 10 Products by Sales
top_product=df.groupby('Product Name')['Sales'].sum()

top_product=top_product.sort_values(ascending=False).head(10)

print(top_product)
```

```python
top_product.plot(kind='bar')

plt.title('Top 10 Product Sales')
plt.xlabel('Product Name')
plt.ylabel('Sales')
plt.show()                 
```

```python
category_sales=df.groupby('Category')['Sales'].sum()
print(category_sales)
```

```python
category_sales.plot(kind='bar')

plt.title('Category Wise Sales')
plt.xlabel('Category')
plt.ylabel('Sales')
plt.show()
```

```python
#Geography-wise Sales
ge_sales=df.groupby('Geography')['Sales'].sum()
print(ge_sales)
```

```python
ge_sales.plot(kind='bar')
plt.title("Geography Wise Sales")
plt.xlabel("Geography")
plt.ylabel("Sales")
plt.show()
```

```python
top_profit=df.groupby('Product Name')['Profit'].sum()

top_profit=top_profit.sort_values(ascending=False).head(10)

print(top_profit)
```

```python
top_profit.plot(kind='bar')

plt.title('Top 10 Profitable Products')
plt.xlabel('Product Name')
plt.ylabel('Profit')
plt.show()          
```

```python
#Monthly Sales Trend

#Convert date column
df['Order Date']=pd.to_datetime(df['Order Date'])
print(df['Order Date'])
```

```python
#Create Month column:
df['Month']=df['Order Date'].dt.month_name()
print(df["Month"])
```

```python
#Monthly sales
monthly_sales=df.groupby('Month')['Sales'].sum()
print(monthly_sales)
```

```python
monthly_sales.plot(kind='line' , marker=0)
plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.show()
```

```python
#Correlation Analysis
corr=df[['Sales','Quantity','Profit']].corr()
print(corr)
```

```python
#heatmap
sns.heatmap(corr,annot=True)
plt.title("Correlation Matrix")
plt.show()
```
